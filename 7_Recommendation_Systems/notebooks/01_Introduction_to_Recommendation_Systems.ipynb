```python
#Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#Create a sample movie dataset
movies = pd.DataFrame({
    "movie_id": [1, 2, 3, 4, 5, 6, 7, 8],

    "title": [
        "Avengers",
        "Iron Man",
        "Titanic",
        "The Notebook",
        "Interstellar",
        "Gravity",
        "John Wick",
        "The Dark Knight"
    ],

    "genres": [
        "Action Sci-Fi Superhero",
        "Action Sci-Fi Superhero",
        "Romance Drama",
        "Romance Drama",
        "Sci-Fi Drama",
        "Sci-Fi Adventure",
        "Action Thriller",
        "Action Crime Thriller"
    ]
})

display(movies)
```

```python
#Convert movie genres into numerical vectors
#We will use TF-IDF to convert movie genres into numerical representations
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(
    movies["genres"]
)

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
#Inspect the vocabulary
print(tfidf.get_feature_names_out())
```

```python
#Inspect numerical representation
tfidf_array = tfidf_matrix.toarray()

tfidf_dataframe = pd.DataFrame(
    tfidf_array,
    columns=tfidf.get_feature_names_out(),
    index=movies["title"]
)

display(tfidf_dataframe.round(2))
```

```python
#Calculate Cosine Similarity
#calculating similarity between all movies
similarity_matrix = cosine_similarity(
    tfidf_matrix
)

print("Similarity Matrix Shape:", similarity_matrix.shape)
```

```python
#Display similarity matrix
similarity_df = pd.DataFrame(
    similarity_matrix,
    index=movies["title"],
    columns=movies["title"]
)

display(similarity_df.round(2))
```

```python
#Building my first movie recommendation function
def recommend_movies(movie_title, number_of_recommendations=3):

    # Find the movie index
    movie_index = movies[
        movies["title"] == movie_title
    ].index[0]

    # Get similarity scores for the selected movie
    similarity_scores = list(
        enumerate(similarity_matrix[movie_index])
    )

    # Sort movies by similarity
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the movie itself
    similarity_scores = similarity_scores[1:]

    # Select top recommendations
    top_movies = similarity_scores[
        :number_of_recommendations
    ]

    print(f"Recommendations for: {movie_title}")

    for index, score in top_movies:

        print(
            movies.iloc[index]["title"],
            "| Similarity:",
            round(score, 3)
        )
```

```python
#Test your recommendation engine
recommend_movies("Avengers", 3)
recommend_movies("Titanic", 3)
recommend_movies("John Wick", 3)
```

```python
#Collaborative Filtering
ratings = pd.DataFrame({
    "Avengers": [5, 5, 1, 2],
    "Iron Man": [5, 4, 1, 1],
    "Titanic": [1, 1, 5, 5],
    "Interstellar": [4, 5, 2, 1]
}, index=["User A", "User B", "User C", "User D"])

display(ratings)
```

```python
#User-Based Collaborative Filtering
user_similarity = cosine_similarity(ratings)

user_similarity_df = pd.DataFrame(
    user_similarity,
    index=ratings.index,
    columns=ratings.index
)

display(user_similarity_df.round(2))
```

```python
#Find similar users
selected_user = "User A"

similar_users = user_similarity_df[
    selected_user
].drop(selected_user)

similar_users = similar_users.sort_values(
    ascending=False
)

print("Users most similar to", selected_user)

display(similar_users)
```
