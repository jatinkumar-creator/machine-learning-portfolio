# Day 4 — Introduction to Data Preprocessing

## Objectives
- Understand data preprocessing.
- Identify numerical and categorical data.
- Inspect a dataset using Pandas.
- Identify missing values and duplicates.
- Understand why cleaning is important.

```python
#import libraries
import pandas as pd
import numpy as np
```

```python
#sample dataset
data = {
    "Name": ["Rahul", "Amit", "Priya", "Rahul", "Neha"],
    "Age": [25, np.nan, 30, 25, 200],
    "Salary": [30000, 45000, np.nan, 30000, 50000],
    "Purchased": ["Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

df
```

```python
#inspect the dataset
print("First five rows:")
display(df.head())

print("Dataset shape:")
print(df.shape)

print("Column information:")
df.info()

print("Statistical summary:")
display(df.describe(include="all"))
```

```python
#find missing values
print(df.isnull())

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
#find duplicate records
print("Duplicate rows:")
display(df[df.duplicated()])

print("Total duplicate rows:")
print(df.duplicated().sum())
```

# Day 5 — Handling Missing Values

## Objectives
- Identify missing values.
- Understand deletion techniques.
- Understand mean, median and mode imputation.
- Implement SimpleImputer using Scikit-learn.

```python
#Create a dataset
import pandas as pd
import numpy as np

data = {
    "Age": [25, 30, np.nan, 40, 35, np.nan, 28],
    "Salary": [30000, 40000, 50000, np.nan, 45000, 60000, 35000],
    "City": ["Delhi", "Mumbai", "Delhi", np.nan, "Mumbai", "Delhi", "Delhi"]
}

df = pd.DataFrame(data)

display(df)
```

```python
#Check missing values
print("Missing values:")
print(df.isnull().sum())
```

```python
#Calculate mean, median and mode
print("Age mean:", df["Age"].mean())

print("Age median:", df["Age"].median())

print("City mode:", df["City"].mode()[0])
```

```python
#Apply imputation using Pandas
df_filled = df.copy()

df_filled["Age"] = df_filled["Age"].fillna(
    df_filled["Age"].median()
)

df_filled["Salary"] = df_filled["Salary"].fillna(
    df_filled["Salary"].mean()
)

df_filled["City"] = df_filled["City"].fillna(
    df_filled["City"].mode()[0]
)

display(df_filled)
```

```python
#Verify the result
print("Missing values before:", df.isnull().sum().sum())

print("Missing values after:", df_filled.isnull().sum().sum())
```
